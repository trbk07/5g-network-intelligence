All results below use validation only. Do not use holdout for selection.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Markdown, display
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
skill = pd.read_csv(root / 'reports/lightgbm_skill_val.csv')
display(skill[['model', 'horizon', 'mae', 'best_baseline_mae', 'skill_score', 'best_iteration']])
skill.pivot(index='horizon', columns='model', values='mae').plot.bar()
plt.ylabel('Validation MAE - DLPRB usage')
plt.tight_layout()
plt.show()
```

```python
importance = pd.read_csv(root / 'reports/lightgbm_feature_importance.csv')
top = importance[(importance.model == 'lightgbm_c1') & (importance.horizon == 1)].nlargest(15, 'gain')
top.sort_values('gain').plot.barh(x='feature', y='gain', legend=False)
plt.title('C1 +1h feature gain - descriptive importance')
plt.tight_layout()
plt.show()
display(pd.read_csv(root / 'reports/lightgbm_failure_examples.csv'))
display(Markdown((root / 'reports/lightgbm_analysis.md').read_text(encoding='utf-8')))
```
