# Week 4 GRU-A/B validation comparison
Run GPU training and `python -m scripts.evaluate_models --model gru --segment val` first.
This notebook reads validation artifacts only. Mean and SD across two seeds are descriptive.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Markdown, display
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
comparison = pd.read_csv(root / 'reports/gru_comparison_val.csv')
display(comparison)
comparison.pivot(index='horizon', columns='variant', values='mae_mean').plot.bar()
plt.ylabel('Validation MAE - DLPRB usage')
plt.title('GRU-A/B means across seeds')
plt.tight_layout()
plt.show()
display(pd.read_csv(root / 'reports/gru_benchmark_val.csv'))
display(Markdown((root / 'reports/gru_error_analysis.md').read_text(encoding='utf-8')))
```
